```python

def credit_decision(dti, credit_score):
    if credit_score < 620:
        decision = "Decline"
    elif dti > 0.43:
        decision = "Decline"
    elif dti > 0.36 or credit_score < 680:
        decision = "Manual review"
    else:
        decision = "Approve"
    return decision

dti = .5
credit_score = 720

print(credit_decision(dti, credit_score))
```

```python
def monthly_payment(principal, annual_rate, years):
    """Return the fixed monthly payment on a fully amortizing loan"""
    r = annual_rate / 12
    n = years * 12
    if r == 0:
        return (principal)/n
    return (principal * r)/(1-(1+r)**-n)

payment = monthly_payment(24_000, 0, 5)

print(round(payment, 2))
```

```python
def amortization_schedule(principal, annual_rate, years, extra_payment=0):
    """Return a month-by-month amortization schedule as a list of dicts."""
    pay = monthly_payment(principal, annual_rate, years)
    balance = principal
    schedule = []
    r = annual_rate / 12
    n = years * 12
    month = 0
    while balance > 0 and month < 600:
        month += 1
        real_pay = pay + extra_payment
        interest = r * balance
        if balance + interest < real_pay:
            real_pay = balance + interest
        principal_paid = real_pay - interest
        balance -= principal_paid
        schedule.append({"month": month, "payment": real_pay, "interest": interest, 
                         "principal": principal_paid, "balance": balance})
    return schedule

schedule = amortization_schedule(300_000, 0.065, 30, extra_payment=300)
print(schedule[:3])
print(schedule[-3:])
```

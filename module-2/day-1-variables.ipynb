```python
principal = 10000
rate = 0.07
years = 10
periods = 1
periods_2 = 12

future_value = principal * (1+rate/periods)**(years*periods)
print(f"${future_value:,.2f}")

future_value_2 = principal * (1+rate/periods_2)**(years*periods_2)
print(f"${future_value_2:,.2f}")

diff = future_value_2 - future_value
print(f"${diff:,.2f}")
```

```python
buy_price = 150
current_price = 182
current_price_2 = 210
divs_collected = 3.2
years_2 = 3

price_return = (current_price - buy_price)/buy_price
print(f"{price_return:,.2%}")

div_yield = divs_collected / current_price
print(f"{div_yield:,.2%}")

div_return = divs_collected / buy_price
print(f"{div_return:,.2%}")

total_return = (current_price + divs_collected - buy_price)/buy_price
print(f"{total_return:,.2%}")

cagr = (current_price_2 / buy_price) ** (1/years_2) -1
print(f"{cagr:,.2%}")
```

```python
assumptions = {"rate": 0.08, "initial": -10000, "growth": 0.05, "base_cf": 2500}
cash_flows = [assumptions["initial"]]
npv = 0
for year in range(1, 6):
    cash_flows.append(assumptions["base_cf"] * ((1+assumptions["growth"])** (year-1)))
for t, cf in enumerate(cash_flows):
    pv = cf/((1+assumptions["rate"])**t)
    npv += pv
    print(f"Year {t}: PV = ${pv:,.2f}")
print(f"NPV: ${npv:,.2f}")
```

```python
rates = [0.06, 0.08, 0.10, 0.12]

for r in rates:
    npv = 0
    for t, cf in enumerate(cash_flows):
        pv = cf/((1+r)**t)
        npv += pv
    print(f"Rate: {r}, NPV: ${npv:,.2f}")
```

Excel's NPV(0) treats the first value as period 1, so the year 0 flow has to be added outside the function
